# Active case

One control sets the equity for every model notebook (`01`–`06`).

Charts share a single quant theme (Plotly). Sensitivity surfaces map **steady-state FCFF**, not the raw forecast path.


```python
%matplotlib inline
import sys
from pathlib import Path

ROOT = next(p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (p / "valuation" / "__init__.py").exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import pandas as pd
from valuation import bind_equity_switcher, plots as viz, report as R
from valuation.statements import Line, forecast_statements

viz.style()
```

## Select equity


```python
case = bind_equity_switcher()
UNIT = case.unit
```

## Snapshot


```python
fc = forecast_statements(case)
inc = fc.statements.income
levels = pd.DataFrame({
    "Revenue": inc.loc[Line.REVENUE].astype(float),
    "EBITDA": inc.loc[Line.EBITDA].astype(float),
    "Net income": inc.loc[Line.NET_INCOME].astype(float),
})
viz.plot_desk_series(
    levels,
    title=case.name,
    ylabel=case.unit,
)
R.show_model_snapshot(case)
```
